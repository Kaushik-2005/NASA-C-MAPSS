```python
import numpy as np
from src.foundations.linear_regression import LinearRegressionGD
from src.foundations.logistic_regression import LogisticRegressionGD

# Small deterministic fixture: one normalized sensor trend and cycle-like feature.
X = np.array([[-2.0], [-1.0], [0.0], [1.0], [2.0]])
y_rul = 80.0 - 5.0 * X[:, 0]
y_failure = (y_rul <= 75).astype(int)
```

```python
linear = LinearRegressionGD(learning_rate=0.05, epochs=2000).fit(X, y_rul)
logistic = LogisticRegressionGD(learning_rate=0.5, epochs=2000).fit(X, y_failure)
print(linear.loss_history_[0], linear.loss_history_[-1])
print(logistic.loss_history_[0], logistic.loss_history_[-1])
```
